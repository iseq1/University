## 🎯 **Цель проекта**

Создать **интеллектуальную систему** на любую тему (на выбор студента): машинный перевод, генерация сказок, QA-система, анализ резюме, рекомендации, суммаризация, чат-бот, проверка кода и т.д. Главное – применить современный стек технологий и архитектурные паттерны.

**Система должна включать:**
- Мультиагентную архитектуру (минимум 3 агента + Supervisor)
- Retrieval-Augmented Generation (RAG) с гибридным поиском (BM25 + векторный поиск)
- Два варианта LLM: Ollama и HuggingFace Transformers, а также возможность подключения внешних API (OpenAI, Anthropic, Google и др.)
- REST API (FastAPI)
- Асинхронную обработку, кэширование, фоновые задачи (Celery)
- Human-in-the-Loop (обратная связь и обработка неоднозначных результатов)
- **Систему оценки качества: вычисление метрик на выборках 20, 50, 100 образцов, сравнение моделей и режимов работы**
- **Сквозное логирование шагов рассуждений (reasoning log) и мониторинг**
- Контейнеризацию (Docker)
- Тестирование (pytest)
- Простой UI (Streamlit или аналог)

**Важно:** студенты **не** используют готовые фреймворки для агентов (LangChain, LangGraph и т.п.). Вся логика агентов и оркестрации реализуется самостоятельно. Готовый код не предоставляется, даётся только описание архитектуры и требований.

## 🤖 **Требования к мультиагентной системе**

1. **Базовый класс агента**
   - Абстрактный класс `Agent` с методом `async def process(state) -> state`.
   - Хранит имя агента и последний результат.
   - Все агенты наследуются от него.

2. **Валидатор (Validator)**
   - Выполняет проверку результата по формальным правилам (минимальная длина, наличие обязательных полей, отсутствие повторов и т.п.).
   - Может вызывать LLM для дополнительной проверки.
   - Возвращает список ошибок (`errors`) и предупреждений (`warnings`), а также флаг `passed`.

3. **Критик (Critic)**
   - LLM-as-a-Judge: оценивает результат по нескольким критериям (например, релевантность, полнота, стиль, точность).
   - Использует few-shot prompting с примерами хороших и плохих ответов.
   - Возвращает баллы (0–10) по каждому критерию, общий балл и рекомендации.
   - В случае ошибки LLM (таймаут, неверный JSON) возвращает fallback-оценку (например, 5.0).

4. **Редактор (Editor)**
   - Принимает ошибки/предупреждения от Validator и рекомендации от Critic.
   - Исправляет или улучшает результат через LLM.
   - Возвращает исправленный результат и флаг, изменилось ли что-то.

5. **Супервизор (Supervisor)**
   - Оркеструет работу агентов в цикле:
     ```
     Validator -> Critic -> (если нужно) Editor -> Validator -> Critic -> ...
     ```
   - Ограничивает количество итераций (`max_iterations`).
   - Определяет критерий успеха (например, `validation_passed == True` и `critic_score >= 7`).
   - Ведёт детальный лог всех шагов (`reasoning_log`), который включает:
     - название агента, вызванного на каждом шаге;
     - входные данные агента (или их краткое описание);
     - выходные данные агента (оценки, ошибки, исправленный текст);
     - временные метки начала и окончания выполнения;
     - использованную модель LLM и её параметры (если применялась);
     - возникшие ошибки или предупреждения.
   - Лог должен сохраняться в БД (или файле) и быть доступен через API для просмотра.

## 🔍 **Требования к RAG**

1. **Гибридный поиск**
   - Реализовать два индекса: BM25 (лексический) и векторный (семантический, Qdrant).
   - Объединить результаты методом Reciprocal Rank Fusion (RRF).

2. **Эмбеддинги**
   - Использовать `sentence-transformers` (например, `paraphrase-multilingual-MiniLM-L12-v2`).
   - Оборачивать в класс-синглтон с ленивой загрузкой.

3. **Векторная БД**
   - Qdrant (или аналог: ChromaDB, FAISS) с коллекцией и cosine-метрикой.
   - Добавление, поиск с фильтрами по категориям/языкам (если применимо).

4. **Интеграция с LLM**
   - Контекст из найденных записей подаётся в промпт генеративной модели.

## 🧠 **Поддержка LLM: Ollama, HuggingFace и внешние API**

1. **Ollama** — основной вариант для локального запуска.
   - HTTP API (`/api/generate`).
   - Поддержка разных моделей (например, qwen2.5, mistral, llama3.1).
   - Настройка `temperature` для разных агентов.
   - **Требование к размеру модели**: для локального запуска рекомендуется выбирать модели с числом параметров от **1 до 13 миллиардов** (например, qwen2.5:1.5b, mistral:7b, llama3.1:8b). Это обеспечивает разумное потребление ресурсов и воспроизводимость. В отчёте необходимо указать точные названия и размеры использованных моделей.

2. **HuggingFace Transformers** — запасной вариант для случаев, когда нужна конкретная модель или когда Ollama недоступен.
   - Асинхронная обёртка над `transformers.pipeline`.
   - Кэширование результатов.
   - **Требование к размеру модели**: использовать небольшие модели, пригодные для CPU/одного GPU (например, `google/flan-t5-small`, `distilgpt2`, `microsoft/DialoGPT-small`). Допускается использование более крупных моделей при наличии ресурсов, но это должно быть обосновано.

3. **Внешние API**:
   - Поддержка как минимум двух внешних провайдеров: OpenAI, Anthropic, Google (Gemini), Cohere, Together AI, Groq или OpenRouter.
   - Реализация клиентов в `data/api_llm.py`.
   - Унифицированный интерфейс: `async generate(prompt, provider, **kwargs)`.
   - Обработка ошибок, таймаутов, повторных попыток.

4. **Выбор бэкенда** (через `LLMManager` или аналогичный класс)
   - Если задача поддерживается Ollama — использовать его.
   - Иначе переключаться на HuggingFace или внешний API.
   - Возможность явно указать провайдера через конфигурацию или параметр запроса.

## 📊 **Оценка качества и сравнение моделей**

### 1. **Метрики качества**

Студенты должны реализовать вычисление не менее 5 метрик, подходящих для выбранной задачи. Обязательные для текстовых задач:

- BLEU
- ROUGE-L
- METEOR
- chrF
- BERTScore
- Exact Match (для коротких ответов)

Для задач классификации/рекомендаций можно использовать Accuracy, Precision, Recall, F1, MRR, NDCG.

Все метрики реализуются в `core/metrics.py` или `evaluation/metrics_calculator.py`. Допускается использование готовых библиотек (`sacrebleu`, `rouge-score`, `nltk`, `bert-score`), но студент должен понимать их принцип.

### 2. **Скрипт оценки (`scripts/run_eval.py`)**

Скрипт должен уметь:

- Загружать эталонные данные (CSV/JSON).
- Случайно или стратифицированно выбирать выборки размера **20, 50, 100** образцов.
- Запускать оценку в различных режимах:
  - `baseline` (только LLM без дополнительных компонентов)
  - `+dict` (LLM + справочник)
  - `+rag` (LLM + гибридный поиск)
  - `+agents` (LLM + мультиагентная система)
  - `+ensemble` (ансамбль моделей или методов)
- Оценивать несколько моделей (локальных и внешних API).
- Вычислять метрики и сохранять результаты в JSON/CSV, а также выводить сводную таблицу.
- Опционально: проверять статистическую значимость различий (bootstrap, paired t-test).

Пример вызова:
```bash
python scripts/run_eval.py --data data/test.json --sample-sizes 20 50 100 --modes baseline +rag +agents --models qwen2.5 mistral --output results/
```

### 3. **API для оценки**

Добавить эндпоинты:

- `POST /api/v1/evaluate` — запустить оценку для одной модели и одного режима.
- `GET /api/v1/evaluate/compare?models=...&sample_size=...` — сравнить несколько моделей.
- `POST /api/v1/evaluate/batch` — пакетная оценка (несколько размеров, режимов, моделей).
- `GET /api/v1/evaluate/history` — получить историю сохранённых результатов.

Ответы должны содержать:
- название модели/режима
- размер выборки
- значения метрик
- время вычисления
- timestamp

### 4. **Хранение результатов**

Создать таблицы в БД:
- `evaluation_results`: id, model_name, sample_size, mode, metrics_json, evaluation_time, created_at.
- `model_comparisons`: id, models_compared, sample_size, results_json, winner, created_at.

### 5. **Интеграция с Celery**

Реализовать фоновые задачи:
- `evaluate_model_task` — оценка одной модели.
- `compare_models_task` — сравнение нескольких моделей.
- Задачи должны обновлять статус и сохранять результаты в БД.

### 6. **Визуализация (в UI)**

Добавить страницу в Streamlit:
- Форма для запуска оценки (выбор моделей, размеров, режимов).
- Отображение таблиц сравнения.
- Графики (bar chart, radar chart) с использованием Plotly.

**Задача:**

**Агенты:**
- **Generator** – создаёт черновик сказки по параметрам.
- **Validator** – проверяет формальные требования: объём, наличие всех заданных элементов, отсутствие повторов и несоответствий (например, герой не меняется по ходу сюжета).
- **Critic** – оценивает художественность, логичность, соответствие возрасту, наличие морали (LLM-as-a-Judge).
- **Editor** – исправляет замечания Validator и Critic, улучшает стиль и связность.

**База знаний (RAG):**  
Коллекция русских народных и авторских сказок (1000+ текстов). Справочник – частотные сказочные обороты, имена персонажей, волшебные предметы (500+ записей). Гибридный поиск помогает подбирать похожие сюжеты и стилистические приёмы.

**Данные:**  
Открытые датасеты сказок (например, сборники на GitHub, Lib.ru). Легко набрать более 1000 текстов.

**Метрики:**  
- **BLEU** – сравнение с эталонными сказками (n-граммное совпадение).  
- **ChrF++** – устойчив к перефразированию, учитывает символьные и словесные n-граммы.

- **BERTScore** – семантическая близость на основе эмбеддингов.  
- **ROUGE-L** – покрытие длинных общих подпоследовательностей (дополнительно).  
- **Специфические:** наличие морали (отдельный классификатор), лексическое разнообразие (TTR, distinct-n).

**HITL:**  
Если Critic даёт низкий балл за логичность или стиль, сказка попадает в очередь. Редактор (человек) может переписать фрагмент, после чего пример сохраняется в базу знаний как положительный образец.

### 2. 📐 Решатель математических задач (ОГЭ/ЕГЭ) с пошаговым объяснением

**Задача:**  
Пользователь вводит текст задачи, система выдаёт решение с подробными шагами и ответом.

**Агенты:**
- **Solver** – генерирует пошаговое решение.
- **Validator** – проверяет формальную корректность: наличие всех шагов, корректность математических выражений, отсутствие синтаксических ошибок в LaTeX.
- **Critic** – оценивает правильность каждого шага, логику, полноту объяснения (может сверяться с базой решённых задач).
- **Editor** – исправляет ошибки, добавляет пропущенные шаги, улучшает объяснение.

**База знаний:**  
Набор задач с эталонными решениями (1000+ пар «задача – решение»). Справочник – формулы, теоремы, определения (500+). RAG находит похожие задачи и методы решения.

**Данные:**  
Открытые базы задач ОГЭ/ЕГЭ (например, с сайтов ФИПИ, решу ЕГЭ). Можно спарсить или взять готовые CSV.

**Метрики:**  
- **Exact Match** – точное совпадение конечного ответа (обязательная, если ответ формализуем).  
- **BLEU, ChrF++, METEOR, BERTScore** – для оценки текста пошагового объяснения (после нормализации LaTeX).  
- **ROUGE-L** – для полноты объяснения.  
- **Accuracy по шагам** – сверка каждого шага с эталонным решением (дополнительная структурная метрика).

**HITL:**  
Если решение расходится с эталонным ответом или Critic даёт низкую оценку, пример помечается как неоднозначный. Преподаватель/редактор может подтвердить правильное решение или исправить ошибки, после чего оно добавляется в базу.

### 3. ❓ Генератор QA-пар из текстов (Википедия/новости)

**Задача:**  
Из произвольного текста (абзаца или статьи) автоматически создаются вопросы и ответы (формат Yes/No или фактографические), пригодные для обучения моделей или тестирования.

**Агенты:**
- **QuestionGenerator** – создаёт вопросы по содержанию абзаца.
- **Validator** – проверяет, что вопрос корректен, ответ однозначен и действительно содержится в тексте.
- **Critic** – оценивает уместность, сложность, грамматическую правильность вопроса (LLM-as-a-Judge).
- **Editor** – переформулирует неудачные вопросы, исправляет ошибки.

**База знаний:**  
Набор текстов (статьи Википедии, новости), разбитых на абзацы (1000+). Справочник – частотные сущности, термины, шаблоны вопросов (500+). RAG помогает находить похожие тексты и уже существующие QA-пары для обучения генератора.

**Данные:**  
Дампы Википедии (можно взять готовые выжимки) или новостные датасеты (например, CNN/Daily Mail, Lenta.ru). Легко подготовить 1000+ абзацев.

**Метрики:**  
- **BLEU, ChrF++, METEOR, BERTScore** – для сравнения сгенерированного вопроса с эталонным (текстовое качество).

- **F1** – оценка фактической точности (совпадение ключевых сущностей).  
- **ROUGE-L** – дополнительно для вопроса.

**HITL:**  
Если Validator не может однозначно определить ответ (например, вопрос двусмысленный) или Critic ставит низкий балл, пара «вопрос-ответ» уходит на ручную проверку. Редактор может принять, отклонить или исправить, после чего она пополняет базу.

### 4. 📰 Суммаризатор новостей с оценкой качества

**Задача:**  
Пользователь вводит текст новости, система генерирует краткое резюме (1–3 предложения), сохраняя ключевые факты.

**Агенты:**
- **Summarizer** – создаёт первичное резюме.
- **Validator** – проверяет формальные требования: длина, отсутствие копирования целых предложений из оригинала, наличие ключевых сущностей.
- **Critic** – оценивает полноту (не потеряны ли важные факты) и точность (нет ли искажений), сравнивая с исходным текстом.
- **Editor** – улучшает резюме на основе замечаний.

**База знаний:**  
Набор пар «новость – эталонное резюме» (1000+). Справочник – частые имена, организации, термины (500+). RAG подбирает похожие новости и способы их суммаризации.

**Данные:**  
Датасеты суммаризации (XSum, CNN/Daily Mail) или собранные с новостных сайтов.

**Метрики:**  
- **BLEU** (обычно BLEU-1/2) – для поверхностного сходства.  
- **ChrF++** – рекомендуется для суммаризации, менее чувствителен к длине.  
- **METEOR** – синонимы и перефразирование.  
- **BERTScore** – семантическая эквивалентность.  
- **ROUGE-1/2/L** – стандарт для суммаризации (можно добавить, но не обязательно, если уже есть вышеперечисленные).

**HITL:**  
Если Critic считает, что потерян ключевой факт или допущена фактическая ошибка, резюме попадает в очередь. Редактор исправляет, а исправленный вариант сохраняется как новый эталон.

### 6. 📄 Суммаризатор научных статей (PDF) с оценкой качества

**Задача:**  
Пользователь загружает PDF научной статьи или указывает её идентификатор (например, arXiv ID). Система извлекает текст, генерирует краткое резюме (abstract-like) или развёрнутое резюме по секциям, сохраняя ключевые выводы и методологию.

**Агенты:**
- **Extractor** (опционально) – извлекает текст из PDF, очищает от артефактов, разбивает на секции (Introduction, Methods, Results, Conclusion).
- **Summarizer** – создаёт первичное резюме всей статьи или отдельных секций.
- **Validator** – проверяет формальные требования: длина резюме, наличие ключевых терминов из статьи, отсутствие плагиата (дословного копирования целых предложений).
- **Critic** – оценивает полноту (отражены ли основные результаты и выводы), точность (нет ли искажений фактов), связность и научный стиль (LLM-as-a-Judge).
- **Editor** – исправляет недостатки, улучшает формулировки, добавляет пропущенные детали.

**База знаний (RAG):**  
Коллекция научных статей с эталонными резюме (abstracts) — не менее 1000 пар «полный текст (или ключевые секции) → аннотация».  
Справочник — частотные научные термины, аббревиатуры, шаблонные фразы (например, "We propose a novel method...") — 500+ записей.  
Гибридный поиск помогает находить похожие статьи и способы их суммаризации, а также уточнять значение терминов.

**Данные:**  
- **arXiv** – открытый репозиторий, есть готовые датасеты (например, arXiv Dataset на Kaggle) с полными текстами и аннотациями.  
- **PubMed Central** – биомедицинские статьи в открытом доступе.  
- **Semantic Scholar API** – можно получить аннотации и полные тексты (для открытых статей).  
- Сбор 1000+ статей в PDF или XML вполне реален, но потребуется скрипт для извлечения текста (например, `pdfplumber`, `PyMuPDF`, `GROBID` для научных статей).  
- Для простоты можно взять уже извлечённые тексты из датасетов (CSV/JSONL), не работая напрямую с PDF.

**Метрики:**  
- **BLEU** – n-граммное совпадение (обычно BLEU-1/2).  
- **ChrF++** – устойчив к длинным техническим текстам.  
- **METEOR** – учитывает синонимы научной лексики.  
- **BERTScore** – семантическая близость.  
- **ROUGE-1/2/L** – стандарт для суммаризации.  
- **Coverage Score** – доля ключевых терминов статьи, попавших в резюме (дополнительно).  
- **FactCC** (опционально) – фактическая точность.

**HITL:**  
Если Critic даёт низкий балл за фактическую точность или Validator обнаруживает, что резюме слишком короткое/неполное, статья попадает в очередь.  
Редактор (человек, возможно эксперт) просматривает резюме, может исправить ошибки или дополнить. Подтверждённое резюме сохраняется в базу знаний как новый эталон для будущих генераций.

**Особенности реализации:**
- Для обработки длинных статей потребуется стратегия chunking (разбиение на секции) и иерархическая суммаризация (сначала по секциям, затем общая).  
- Можно добавить режим «краткое резюме» (как abstract) и «подробное резюме» (по секциям).  
- RAG здесь особенно полезен: при генерации резюме можно подтягивать похожие статьи и их аннотации как few-shot примеры или для проверки терминологии.

### 8. 🍳 Генератор кулинарных рецептов по ингредиентам и диете

- **Critic** – оценивает сочетаемость продуктов, реалистичность приготовления, вкусовые качества (LLM-as-a-Judge).

**База знаний:**

### 15. 💻 Ревьюер кода с рекомендациями по улучшению

**Задача:**  
Пользователь загружает фрагмент кода (Python, JavaScript и др.) и описание задачи. Система анализирует код: находит ошибки, потенциальные уязвимости, неэффективные места, даёт рекомендации по рефакторингу и стилю.

**Агенты:**
- **CodeAnalyzer** – первичный анализ кода (синтаксис, типичные ошибки).
- **Validator** – проверяет формальную корректность рекомендаций: наличие пояснений, отсутствие ложных срабатываний, соответствие стандартам.
- **Critic** – оценивает полезность, точность, глубину анализа, читаемость рекомендаций (LLM-as-a-Judge).
- **Editor** – уточняет рекомендации, добавляет примеры исправленного кода, исправляет неточности.

**База знаний:**  
Коллекция пар «код с проблемами → исправленный код + пояснение» (1000+ примеров). Справочник – типовые ошибки, паттерны, правила стиля (500+). RAG находит похожие случаи для обоснования рекомендаций.

Открытые датасеты с code review (GitHub, Stack Overflow, CodeNet, MBPP). Можно собрать 1000+ примеров.

**Метрики:**  
- **BLEU, ChrF++, METEOR, BERTScore** – для оценки текстовых рекомендаций и исправленного кода (при нормализации).  
- **Exact Match** – для проверки конкретных исправлений (например, замены строки).  
- **Precision/Recall/F1** – точность обнаружения реальных проблем (если есть разметка).  
- **Специфические:** CodeBLEU (если реализуемо).

**HITL:**  
Если Critic считает рекомендацию неверной или Validator обнаруживает ложное срабатывание, пример попадает в очередь. Разработчик-редактор подтверждает или исправляет, после чего пара сохраняется в базе.
